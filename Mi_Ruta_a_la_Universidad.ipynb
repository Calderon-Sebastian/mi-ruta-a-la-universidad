```python
Codigo = input("Digite el codigo de usuario: ").strip().upper()
#El codigo para acceder en este caso es "A00435358".

if Codigo == "A00435358":
    print(" ")
    print("Bienvenido(a) al Sistema “Mi Ruta a la Universidad”.\n")
    print("Seleccione una opción:")
    print("--------------------------------")
    print("1. Calcular el costo estimado del desplazamiento")
    print("2. Analizar las características del desplazamiento")
    print("3. Recomendar un medio de transporte")
    print("4. Determinar si llegará a tiempo a la universidad")
    print("5. Estimar el impacto ambiental del desplazamiento")
    print("6. Salir")
    print("--------------------------------\n")

    Opcion = int(input("Ingrese el número de la opción deseada: "))
    print(" ")

    if Opcion < 1 or Opcion > 6:
        print("Error: La opción ingresada no es válida. Por favor, seleccione un número entre 1 y 6.")
        exit()

    def Costo_Desplazamiento():
        #Calcula el valor aproximado de desplazamiento desde el lugar de residencia hasta la univerisadad.

        Tarifa_Base = float(input("Ingrese la tarifa base del transporte: "))
        Distancia = float(input("Ingrese la distancia en kilometros desde su lugar de residencia hasta la universidad: "))
        Costo_Kilometro = float(input("Ingrese el costo por kilometro del transporte: "))
        Hora_pico = int(input("El desplazamiento se realizará en hora pico? (Si: 1/No: 0): "))

        if Tarifa_Base < 0 or Distancia < 0 or Costo_Kilometro < 0:
            print("Error: Los valores ingresados no pueden ser negativos.")
            return

        if Hora_pico not in [0, 1]:
            print("Error: La opción de hora pico debe ser 0 o 1.")
            return

        Cargo_Adicional = 0.0
        if Hora_pico == 1:
            Cargo_Adicional = float(input("Ingrese el cargo adicional por hora pico: "))

        if Cargo_Adicional < 0:
            print("Error: El cargo adicional no puede ser negativo.")
            return

        Costo_Total = Tarifa_Base + (Costo_Kilometro * Distancia) + (Cargo_Adicional * Hora_pico)

        print(" ")
        print(f"El costo estimado del desplazamiento es: ${Costo_Total:.0f}")

    def Caracteristicas_Desplazamiento():
        #Se analiza el recorrido desde el lugar de residencia hasta la universidad.

        Distancia = float(input("Ingrese la distancia en kilometros desde su lugar de residencia hasta la universidad: "))
        Tiempo_Estimado = float(input("Ingrese el tiempo estimado de desplazamiento en minutos: "))

        if Distancia < 0 or Tiempo_Estimado < 0:
            print("Error: Los valores ingresados no pueden ser negativos.")
            return

        if Distancia <= 2:
            msgDistancia = "El trayecto es cercano."
        elif Distancia > 2 and Distancia <= 8:
            msgDistancia = "El trayecto es intermedio."
        else:
            msgDistancia = "El trayecto es lejano."

        if Tiempo_Estimado < 30:
            msgTiempo = "El desplazamiento es rapido."
        elif Tiempo_Estimado <= 60:
            msgTiempo = "El desplazamiento es moderado."
        else:
            msgTiempo = "El desplazamiento es prolongado."

        print (" ")
        print(f"Distancia: {msgDistancia}")
        print(f"Tiempo estimado: {msgTiempo}\n")

        if Distancia <=2 and Tiempo_Estimado > 30:
            print("El tiempo de desplazamiento es alto en relación con la distancia. Conviene revisar la ruta utilizada.")

    def Recomendar_Transporte():
        #Se recomendara una alternativa de trasnporte para llegar a la universidad.

        Distancia = float(input("Ingrese la distancia en kilometros desde su lugar de residencia hasta la universidad: "))
        Lluvia = int(input("Se espera lluvia durante el desplazamiento? (Si: 1/No: 0): "))
        Bicicleta = int(input("Cuenta con bicicleta para el desplazamiento? (Si: 1/No: 0): "))

        if Distancia < 0:
            print("Error: La distancia no puede ser negativa.")
            return

        if Lluvia not in [0, 1]:
            print("Error: La opción de lluvia debe ser 0 o 1.")
            return

        if Bicicleta not in [0, 1]:
            print("Error: La opción de bicicleta debe ser 0 o 1.")
            return

        if Distancia <= 2:
            if Lluvia == 1:
                print("Se recomienda utilizar transporte público")
            else:
                print("Se recomienda caminar")

        elif Distancia > 2 and Distancia <= 8:
            if Bicicleta == 1 and Lluvia == 0:
                print("Se recomienda utilizar bicicleta")
            else:
                print("Se recomienda utilizar transporte público")

        else:
            print("Se recomienda utilizar transporte público o vehiculo compartido.")

    def Llegar_Universidad():
        #Determinara si el estudiante dispone de suficiente tiempo para llegar a tiempo a la universidad.

        Tiempo_Disponible = int(input("Ingrese el tiempo disponible antes de la actividad en minutos: "))
        Tiempo_Estimado = int(input("Ingrese el tiempo estimado de desplazamiento en minutos: "))
        Margen_Seguridad = 10

        if Tiempo_Disponible < 0 or Tiempo_Estimado < 0:
            print("Error: Los valores ingresados no pueden ser negativos.")
            return

        Tiempo_Necesario = Tiempo_Estimado + Margen_Seguridad

        if Tiempo_Disponible >= Tiempo_Necesario:
            print("Puedes realizar el desplazamiento y llegar a tiempo.")
        else:
            print("Existe riesgo de llegar tarde. Se recomienda salir antes o considerar otra alternativa de transporte.")

    def Impacto_Ambiental():
        #Se estima las emisiones de dioxido de carbono (C02) asociadas al desplazamiento a la universidad.

        Distancia = float(input("Ingrese la distancia recorrida en kilometros: "))

        print(" ")
        print ("Seleccione el medio de transporte utilizado: ")
        print("--------------------------------")
        print("1. Caminata")
        print("2. Bicicleta")
        print("3. Transporte público")
        print("4. Motocicleta")
        print("5. Automóvil")
        print("--------------------------------\n")

        Medio_Transporte = int(input("Ingrese el número del medio de transporte utilizado entre 1 y 5: "))

        if Distancia < 0:
            print("Error: La distancia no puede ser negativa.")
            return

        if Medio_Transporte not in [1, 2, 3, 4, 5]:
            print("Error: La opción de medio de transporte debe ser un número entre 1 y 5.")
            return

        if Medio_Transporte == 1:
            Emisiones_CO2 = 0
        elif Medio_Transporte == 2:
            Emisiones_CO2 = 0
        elif Medio_Transporte == 3:
            Emisiones_CO2 = 80
        elif Medio_Transporte == 4:
            Emisiones_CO2 = 100
        else:
            Emisiones_CO2 = 180

        Total_Emisiones_CO2 = Distancia * Emisiones_CO2

        if Total_Emisiones_CO2 == 0:
            Clasificacion = "Sin emisiones directas."
        elif Total_Emisiones_CO2 <= 500:
            Clasificacion = "Impacto bajo."
        elif Total_Emisiones_CO2 > 500 and Total_Emisiones_CO2 <= 1500:
            Clasificacion = "Impacto moderado."
        else:
            Clasificacion = "Impacto alto."

        print(" ")
        print(f"Las emisiones de CO2 asociadas al desplazamiento son: {Total_Emisiones_CO2:.0f}g")
        print(f"Impacto ambiental: {Clasificacion}")

    def Salir():
        #Finaliza el programa y agradece al usuario por utilizar el sistema.
        print("Gracias por utilizar el Sistema “Mi Ruta a la Universidad”. ¡Hasta luego!")
        exit()

    if Opcion == 1:
        Costo_Desplazamiento()

    if Opcion == 2:
        Caracteristicas_Desplazamiento()

    if Opcion == 3:
        Recomendar_Transporte()

    if Opcion == 4:
        Llegar_Universidad()

    if Opcion == 5:
        Impacto_Ambiental()

    if Opcion == 6:
        Salir()

else:
    print("Acceso denegado. Codigo no reconocido.")
```
